```sql
SELECT
  COUNT(*) AS total_rows
FROM
  read_files(
    '/Volumes/olist/bronze/raw_data/olist_orders_dataset.csv',
    format => 'csv',
    header => TRUE
  )
```

```sql
%python
file_path = f"{volume_path}/olist_orders_dataset.csv"

df = spark.read.csv(file_path, header=True)
total_rows = df.count()

print(f"{file_path} -> {total_rows} rows")
```
